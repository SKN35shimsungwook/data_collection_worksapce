### 073
숫자 1이 저장된 튜플을 생성하라.

````python
# 요소가 1개인 튜플을 만들 때는 값 뒤에 반드시 콤마(,)를 붙여야 한다.
# 콤마 없이 (1)만 쓰면 튜플이 아니라 정수 1로 취급된다.
t = (1,)
print(t, type(t))  # (1,) <class 'tuple'>
````

### 074
다음 코드를 실행해보고 오류가 발생하는 원인을 설명하라.
```python
>> t = (1, 2, 3)
>> t[0] = 'a'
Traceback (most recent call last):
  File "<pyshell#46>", line 1, in <module>
    t[0] = 'a'
TypeError: 'tuple' object does not support item assignment
```

````python
# 튜플(tuple)은 리스트와 달리 "변경 불가능(immutable)"한 자료형이다.
# 한 번 생성되면 내부 요소를 추가/삭제/수정할 수 없기 때문에
# 인덱스로 값을 재할당(item assignment)하려고 하면
# TypeError: 'tuple' object does not support item assignment
# 오류가 발생한다.

t = (1, 2, 3)
t[0] = 'a'   # TypeError 발생
````

### 075
아래와 같이 t에는 1, 2, 3, 4 데이터가 바인딩되어 있다. t가 바인딩하는 데이터 타입은 무엇인가?
```python
t = 1, 2, 3, 4
```

````python
# 괄호 없이 콤마로만 값을 나열해도 파이썬은 이를 튜플로 인식한다.
t = 1, 2, 3, 4
print(type(t))  # <class 'tuple'>
````

### 076
변수 t에는 아래와 같은 값이 저장되어 있다. 변수 t가 ('A', 'b', 'c') 튜플을 가리키도록 수정하라.
```python
t = ('a', 'b', 'c')
```

````python
# 튜플 자체는 수정할 수 없으므로(immutable),
# 새로운 튜플을 만들어 변수 t가 그 튜플을 "다시 가리키도록" 재할당한다.
t = ('a', 'b', 'c')
t = ('A', 'b', 'c')
print(t)
````

### 077
다음 튜플을 리스트로 변환하라.
```python
interest = ('삼성전자', 'LG전자', 'SK Hynix')
```

````python
# list() 함수를 사용하면 튜플을 리스트로 변환할 수 있다.
interest = ('삼성전자', 'LG전자', 'SK Hynix')
interest = list(interest)
print(interest, type(interest))  # ['삼성전자', 'LG전자', 'SK Hynix'] <class 'list'>
````

### 078
다음 리스트를 튜플로 변경하라.
```python
interest = ['삼성전자', 'LG전자', 'SK Hynix']
```

````python
# tuple() 함수를 사용하면 리스트를 튜플로 변환할 수 있다.
interest = ['삼성전자', 'LG전자', 'SK Hynix']
interest = tuple(interest)
print(interest, type(interest))  # ('삼성전자', 'LG전자', 'SK Hynix') <class 'tuple'>
````

### 079 튜플 언팩킹
다음 코드의 실행 결과를 예상하라.
```python
temp = ('apple', 'banana', 'cake')
a, b, c = temp
print(a, b, c)
```

````python
# 튜플 언패킹은 튜플의 각 요소를 순서대로 여러 변수에
# 한 번에 나누어 할당하는 기능이다. temp에 요소가 3개이므로
# 변수도 a, b, c 3개로 받아야 한다.
temp = ('apple', 'banana', 'cake')
a, b, c = temp
print(a, b, c)  # apple banana cake
````

### 080 range 함수
1부터 99까지의 정수 중 짝수만 저장된 튜플을 생성하라.

(2, 4, 6, 8 ... 98)

````python
# range(시작, 끝, 증가폭)에서 시작을 2, 끝을 100(미포함이므로 99 다음),
# 증가폭을 2로 주면 2부터 98까지 짝수만 만들어진다.
# 이를 tuple()로 감싸 튜플로 변환한다.
even_numbers = tuple(range(2, 100, 2))
print(even_numbers)  # (2, 4, 6, 8, ..., 98)
````
