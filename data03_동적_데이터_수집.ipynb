
### 시간창 대기 3가지
- 일반적으로 시간창 대기는 time.sleep , implicitly_wait , Explicit Waits 3가지 사용
 - time.sleep: 그냥 지정한 시간만큼 기다린 뒤에 작업을 진행.
 - implicitly_wait : 지정한 시간만큼 기다려주고 그 이후는 기다리지 않음.
 - Explicit Waits : 특정 상태가 될때까지 기다리고, 상태가 되면 바로 다음 실행.

```python
# 옵션 설정
options = Options()
options.add_argument("--start-maximized") # 화면 최대화


# 실행 시 최신 ChromeDriver를 자동으로 다운로드 및 로드합니다.
service = Service(ChromeDriverManager().install())


# 드라이버 실행
driver = webdriver.Chrome(service=service, options=options)
```

```python
"""
서울특별시 장애인 복지 / 서울특별시 장애인 지원 / 서울특별시 장애인 콜택시
1) 네이버 통합검색 진입 → 2) '뉴스' 탭 클릭 → 3) 무한 스크롤로 기사 100개까지 수집
"""

import time
import random
import csv
from datetime import datetime

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, NoSuchElementException
from webdriver_manager.chrome import ChromeDriverManager


KEYWORDS = [
    ("서울특별시 장애인 복지", "복지지원사업"),
    ("서울특별시 장애인 지원", "복지지원사업"),
    ("서울특별시 장애인 콜택시", "장애인콜택시"),
]

TARGET_COUNT = 100          # 키워드당 목표 수집 건수
MAX_SCROLL_NO_CHANGE = 4     # 스크롤해도 새 기사가 안 늘어나는 횟수 한도 (더 이상 없다고 판단)
NEWS_TAB_SELECTOR = (
    "#lnb > div.lnb_group > div > div.lnb_nav_area._nav_area_root > div:nth-child(2) > a"
)
CSV_FIELDS = ["category", "keyword", "title", "press", "url", "pub_date", "crawled_at"]


def build_driver():
    options = Options()
    options.add_argument("--start-maximized")
    options.add_argument("--disable-blink-features=AutomationControlled")
    service = Service(ChromeDriverManager().install())
    return webdriver.Chrome(service=service, options=options)


def go_to_news_tab(driver, keyword):
    """네이버 통합검색 진입 후 '뉴스' 탭 클릭"""
    url = f"https://search.naver.com/search.naver?query={keyword}"
    driver.get(url)

    try:
        news_tab = WebDriverWait(driver, 6).until(
            EC.element_to_be_clickable((By.CSS_SELECTOR, NEWS_TAB_SELECTOR))
        )
        news_tab.click()
        time.sleep(random.uniform(1.5, 2.5))
        return True
    except (TimeoutException, NoSuchElementException):
        print(f"[{keyword}] 뉴스 탭 버튼을 못 찾음 (선택자 확인 필요)")
        return False


def extract_items(driver):
    """현재 로드된 뉴스 기사 블록 리스트 반환 (구조 변경 대비 여러 선택자 시도)"""
    for selector in ["div.news_wrap", "div.news_area", "li.bx"]:
        items = driver.find_elements(By.CSS_SELECTOR, selector)
        if items:
            return items
    return []


def parse_item(item):
    title_tag = None
    for sel in ["a.news_tit", "a.title_link", "a[href^='http']"]:
        try:
            title_tag = item.find_element(By.CSS_SELECTOR, sel)
            if title_tag.text.strip():
                break
        except NoSuchElementException:
            continue
    if title_tag is None:
        return None

    title = title_tag.text.strip()
    link = title_tag.get_attribute("href")
    if not title or not link:
        return None

    try:
        press = item.find_element(By.CSS_SELECTOR, "a.info.press, span.info.press").text.strip()
    except NoSuchElementException:
        press = ""

    try:
        info_spans = item.find_elements(By.CSS_SELECTOR, "span.info")
        pub_date = info_spans[-1].text.strip() if info_spans else ""
    except NoSuchElementException:
        pub_date = ""

    return title, link, press, pub_date


def crawl_keyword(driver, keyword, category, seen_urls, result):
    if not go_to_news_tab(driver, keyword):
        return

    collected = 0
    no_change_count = 0
    last_count = 0

    while collected < TARGET_COUNT and no_change_count < MAX_SCROLL_NO_CHANGE:
        items = extract_items(driver)

        for item in items:
            parsed = parse_item(item)
            if parsed is None:
                continue
            title, link, press, pub_date = parsed

            if link in seen_urls:
                continue

            # 제목에 '장애인'이 포함된 기사만 수집
            if "장애인" not in title:
                continue

            seen_urls.add(link)
            result.append({
                "category": category,
                "keyword": keyword,
                "title": title,
                "press": press,
                "url": link,
                "pub_date": pub_date,
                "crawled_at": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            })
            collected += 1

            if collected >= TARGET_COUNT:
                break

        print(f"[{category}|{keyword}] 누적 수집 {collected}건")

        if collected == last_count:
            no_change_count += 1
        else:
            no_change_count = 0
        last_count = collected

        # 페이지 맨 아래로 스크롤 → 추가 기사 로드 대기
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(random.uniform(1.5, 2.5))

    reason = "목표 건수 도달" if collected >= TARGET_COUNT else "더 이상 새 기사 없음"
    print(f"[{category}|{keyword}] 수집 종료 ({reason}) - 총 {collected}건\n")


def main():
    driver = build_driver()
    seen_urls = set()
    result = []

    try:
        for keyword, category in KEYWORDS:
            print(f"\n=== '{keyword}' ({category}) 수집 시작 ===")
            crawl_keyword(driver, keyword, category, seen_urls, result)
    finally:
        driver.quit()

    filename = f"disability_news_{datetime.now().strftime('%Y%m%d_%H%M%S')}.csv"
    with open(filename, "w", newline="", encoding="utf-8-sig") as f:
        writer = csv.DictWriter(f, fieldnames=CSV_FIELDS)
        writer.writeheader()
        writer.writerows(result)

    print(f"\n총 {len(result)}건 저장 완료 → {filename}")


if __name__ == "__main__":
    main()
```
