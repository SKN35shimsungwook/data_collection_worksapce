## 의료데이터 수집
### 1. **공식 데이터 포털에서 의료 데이터 수집**

한국 또는 글로벌 공공기관에서 제공하는 **공공 의료 데이터셋**을 사용하는 방법

🔹 국내: 한국에서 사용 가능한 대표적인 의료 데이터 출처

| 출처 | 내용 | 링크 |
| --- | --- | --- |
| **보건복지부 OPEN DATA** | 병원 정보, 진료 과목, 보건소, 약국 위치, 감염병 등 | [공공데이터포털](https://www.data.go.kr/search/index.do?index=DATA&query=%EB%B3%B4%EA%B1%B4%EB%B3%B5%EC%A7%80) |
| **건강보험심사평가원 (HIRA)** | 병원별 진료비, 처방 통계, 질병 통계, 약품정보 등 | [HIRA 오픈 데이터](https://opendata.hira.or.kr/) |
| **질병관리청(KDCA)** | 감염병 발생, 예방접종, 역학조사 데이터 | [KDCA 감염병포털](https://www.kdca.go.kr/) |
| **의약품안전나라** | 의약품 허가 정보, 이상사례 보고 등 | [의약품안전나라](https://www.drugsafe.or.kr/) |
| **통계청 KOSIS** | 사망원인통계, 건강통계 등 | [KOSIS 건강](https://kosis.kr/) |

🔹 해외: 대표적인 글로벌 오픈 데이터

| 출처 | 내용 | 링크 |
| --- | --- | --- |
| **WHO (세계보건기구)** | 전세계 보건지표, 감염병 등 | [WHO Data](https://www.who.int/data) |
| **CDC (미국 질병통제센터)** | 질병 및 건강행태 데이터 | [CDC Data](https://data.cdc.gov/) |
| **NIH (국립보건원)** | 연구용 의료데이터 (예: MIMIC) | [NIH](https://www.nih.gov/) |
| **PhysioNet** | MIMIC, ECG, Vital Sign 등 실제 환자 생체 신호 | https://physionet.org/ |

### 2. **크롤링을 통한 실시간 의료 정보 수집**

**주의**: 병원 홈페이지나 정부 웹사이트의 크롤링은 **robots.txt**, **이용약관**을 반드시 확인해야 하며, 크롤링이 금지된 경우 **API 사용** 또는 **공공 데이터 다운로드** 방식을 권장함.

### 3. [선택] **직접 수집한 의료 데이터 예시**

- 병원 예약 데이터
- 진료 기록 (EMR)
- 건강검진 결과
- 웨어러블 기기(Apple Watch, Fitbit 등)의 생체정보
- 환자의 자가 보고(PHQ-9 우울증 테스트 등)

> 민감한 개인정보이므로 의료법, 개인정보보호법, HIPAA 등의 법령을 준수해야함.
