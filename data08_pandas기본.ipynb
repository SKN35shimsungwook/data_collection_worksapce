## 1.3 Pandas 자료구조

분석을 위해 다양한 소스(source)로부터 수집하는 데이터는 형태나 속성이 매우 다양하다. 특히, 서로 다른 형식을 갖는 여러 종류의 데이터를 컴퓨터가 이해할 수 있도록 동일한 형식을 갖는 구조로 통합할 필요가 있다. 판다스의 일차적인 목적은 **형식적으로 서로 다른 여러 가지 유형의 데이터를 공통의 포맷으로 정리**하는 것이다.  

판다스는 **시리즈(Series)와 데이터프레임(DataFrame)이라는 구조화된 데이터 형식**을 제공한다. 서로 다른 종류의 데이터를 한곳에 담는 그릇(컨테이너)이 된다. 다만, 시리즈는 1차원 배열이고, 데이터프레임이 2차원 배열이라는 점에서 차이가 있다. 특히, 행과 열로 이루어진 2차원 구조의 데이터프레임은 데이터 분석 실무에서 자주 사용된다.

```python
import pandas as pd

# 리스트를 사용한 시리즈 생성 (인덱스 직접 지정)
population = pd.Series([940, 330, 290], index=['서울', '부산', '인천'])
print(population)

# index 확인하기
print(population.index)   # Index(['서울', '부산', '인천'], dtype='str')

# 배열 확인하기
print(population.values)   # [940 330 290]
```

```python
# k:v 구조를 갖는 딕셔너리를 만들고, 변수 dict_data에 저장
dict_data = {'a': 1, 'b': 2, 'c': 3}
print(dict_data)            # {'a': 1, 'b': 2, 'c': 3}

# 판다스 Series() 함수로 딕셔너리(dict_data)를 시리즈로 변환. 변수 sr에 저장
sr = pd.Series(dict_data)

# 변수 sr의 자료형 출력
print(type(sr))            # <class 'pandas.core.series.Series'>

# 주피터 환경에서 변수 출력
print(sr)

# 인덱스 배열은 변수 idx에 저장
idx = sr.index
print(idx)

# 데이터 값 배열은 변수 val에 저장
val = sr.values
print(val)
```

```python
import pandas as pd

# 리스트를 시리즈로 변환하여 변수 sr에 저장
list_data = ['2019-01-02', 3.14, 'ABC', 100, True]
sr = pd.Series(list_data)
print(sr)

# 인덱스 배열은 변수 idx에 저장
idx = sr.index
print(idx)                         # RangeIndex(start=0, stop=5, step=1)

# 데이터 값 배열은 변수 val에 저장
val = sr.values
print(val)                         # ['2019-01-02' 3.14 'ABC' 100 True]
```

#### ① `loc[]` — 이름(Label) 기반 선택

- **기본 구조:** `df.loc[행_이름, 열_이름]`

```python
# 1. 단일 행 선택 (인덱스 이름이 0인 행)
df.loc[0]
```

```python
# 2. 여러 행 선택 (0번, 2번 행)
df.loc[[0, 2]]
```

```python
# 3. 행 슬라이싱 (주의: loc 슬라이싱은 '끝 범위'를 포함함! 0부터 2까지)
df.loc[0:2]
```

```python
# 4. 행과 열을 동시에 선택 (0~2번 행의 '이름', '부서' 열)
df.loc[0:2, ['tip', 'smoker']]
```

## 4.3 조건 검색

| **방식** | **작성 예시** | **핵심 특징** |
| --- | --- | --- |
| **비교 연산자** | `df[df['sepal_length'] > 5.0]` | 가장 기본적이고 표준적인 검색 방식 |
| **논리 연산자** | `df[(조건1) & (조건2)]` | 여러 조건 조합 시 **조건마다 괄호 `()` 필수** |
| **`isin()`** | `df[df['species'].isin(['setosa', 'virginica'])]` | 여러 항목 중 포함 여부 검사 (OR 대체) |
| **`between()`** | `df[df['sepal_length'].between(5.0, 6.0)]` | 특정 범위(`a <= x <= b`) 간결하게 표현 |
| **`query()`** | `df.query('sepal_length > 5.0 and species == "setosa"')` | 대괄호 없이 **문자열로 조건 작성** (가독성 최상) |

### 4.3.1 비교 연산자(`>`, `<`, `>=`, `<=`, `==`, `!=`)

특정 컬럼의 값과 조건을 비교하여 **조건에 맞는 행만 필터링한**다.

```python
# 1. 총 지불 금액(total_bill)이 30.0 초과인 데이터 검색
df[df['total_bill'] > 30.0]
```

```python
# 2. 요일(day)이 'Sun'(일요일)인 데이터만 검색
df[df['day'] == 'Sun']
```

### 4.3.2 논리 연산자(`&`, `|`, `~`)

두 개 이상의 조건을 조합할 때 사용합니다.

> ⚠️ **주의:** 파이썬 키워드(`and`, `or`, `not`) 대신 비트 연산자(`&`, `|`, `~`)를 사용해야 하며, **각 조건은 반드시 괄호 `()`로 감싸야 합니다.**
> 
- `&` (AND): 두 조건 모두 참
- `|` (OR): 두 조건 중 하나라도 참
- `~` (NOT): 조건의 반대 (부정)

```python
# AND(&): 총 지불 금액(total_bill)이 30.0 이상이고 요일(day)이 'Sun'인 데이터
df[(df['total_bill'] >= 30.0) & (df['day'] == 'Sun')]
```

```python
# OR(|): 총 지불 금액(total_bill)이 40.0 이상이거나 10.0 이하인 데이터
df[(df['total_bill'] >= 40.0) | (df['total_bill'] <= 10.0)]
```

```python
# NOT(~): 요일(day)이 'Thur'가 아닌 데이터
df[~(df['day'] == 'Thur')]
```

### 4.3.3 isin() — 특정 목록 포함 여부 확인

지정한 **리스트 안의 값들 중 하나라도 일치**하면 선택합니다. (여러 값에 대한 OR 조건문 작성 시 유용)

```python
# 요일(day)이 'Sun' 또는 'Sat' (주말)인 데이터 검색
target_days = ['Sun', 'Sat']
df[df['day'].isin(target_days)]['day'].unique()
```

```python
df['day'].unique()
```

```python
# 반대로 지정 목록에 없는 데이터만 뽑을 때 (~ 붙이기) -> 평일(Thur, Fri) 데이터
df[~df['day'].isin(target_days)]
```

### 4.3.4  between() - 범주(범위)데이터 검색

특정 값의 **최솟값과 최댓값 사이(이상~이하)** 범위에 속하는 데이터를 선택합니다.

- **기본값:** `inclusive='both'` (이상/이하, 양 끝값 포함)

```python
# tip이 2.0 이상 3.0 이하인 데이터 검색
# df[(df['tip'] >= 2.0) & (df['tip'] <= 3.0)] 와 동일
df[df['tip'].between(2.0, 3.0)]

#inclusive: Literal['both', 'neither', 'left', 'right'] = "both”
```

### 4.3.5 query() - 문자열 표현식으로 검색

Pandas 대괄호 `[]`나 복잡한 괄호 `()` 없이 **SQL의 WHERE 절처럼 직관적인 문자열 조건식**으로 필터링할 때 사용다.

```python
# 1. 기본 조건 및 논리 연산
res_query1 = df.query('total_bill > 30.0 and day == "Sun"')

# 2. 파이썬 변수를 조건식에 넣을 때 (@ 기호 사용)
min_tip = 5.0
res_query2 = df.query('tip >= @min_tip')

# 3. in 키워드 조합
res_query3 = df.query('day in ["Sun", "Sat"]')

print(res_query1[['total_bill', 'tip', 'day']].head())
```

## 4.4 정렬

Pandas에서는 데이터를 정렬하기 위해 `sort_values()`와 `sort_index()` 메서드를 사용한다.

| 구분 | sort_values() | sort_index() |
| --- | --- | --- |
| 정렬 기준 | 데이터 값(Value) | 인덱스(Index) |
| 사용 목적 | 특정 컬럼 기준 정렬 | 행 또는 열 인덱스 정렬 |
| 여러 컬럼 정렬 | 가능 | 불가능 |
| 주요 옵션 | by, ascending | axis, ascending |

### 4.4.1 sort_values()

sort_values`()`는 값을 **기준으로 정렬**하는 메서드이다

```python
# 1. 총 지불 금액(total_bill) 오름차순 정렬
df.sort_values(by="total_bill")
```

```python
# 2. 총 지불 금액(total_bill) 내림차순 정렬
df.sort_values(
    by="total_bill",
    ascending=False
)
```

```python
# 3. 팁(tip) 기준 정렬
df.sort_values(by="tip")
```

```python
# 4. 인원수(size) 내림차순 정렬
df.sort_values(
    by="size",
    ascending=False
)
```

```python
# 5. 여러 컬럼 기준 정렬
# 요일(day)은 오름차순, 총 지불 금액(total_bill)은 내림차순으로 정렬한다.
df.sort_values(
    by=["day", "total_bill"],
    ascending=[True, False]
)
```

```python
# 6. 원본 데이터 변경 (inplace=True)
df2 = df.copy()
print(df2.head())

df2.sort_values(
    by="tip",
    inplace=True
)
print(df2.head())
```

### 4.4.2 sort_index()

`sort_index()`는 **인덱스를 기준으로 정렬**하는 메서드이다.

```python
# 1. 인덱스를 섞어보기 (무작위 10개 추출)
sample = df.sample(10, random_state=42)
print(sample.head())
```

```python
# 2. 인덱스 오름차순 정렬
print(sample.sort_index())
```

```python
# 3. 인덱스 내림차순 정렬
print(sample.sort_index(
    ascending=False
))
```

```python
# 4. 열(Column) 이름 정렬
# 열의 순서를 알파벳 순(sex, day, size, time, tip, total_bill 등)으로 정렬한다.
df.sort_index(axis=1)
```
